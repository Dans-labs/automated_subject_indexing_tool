```python
import pandas as pd
```

```python
df = pd.read_csv('data/output/aggregated_keywords_SBERT_0.8.csv')
```

```python
df
```

```python
# Check how many unique DOIs are in the dataframe
print(len(df['DOI'].unique()))
doi_list = df['DOI'].unique().tolist()
print(doi_list[0])
print(len(doi_list))
```

```python
# Remove the prefix so that we can compare with the txt file
clean_doi_list = []
for d in doi_list: 
    d = d.split('doi:10.17026/')[1]
    clean_doi_list.append(d)

print(clean_doi_list[0])
```

```python
# read txt file 
with open('data/dois_for_asi_test1.txt', 'r') as f:
    doi_txt_list = f.read().splitlines()
```

```python
# There are 60 DOIS in the txt file, but only 57 in the output csv 
print(doi_txt_list[0])
print(len(doi_txt_list))
```

```python
# Find how many DOIs are in both lists
intersect = set(clean_doi_list).intersection(set(doi_txt_list))
print(len(intersect))
```

```python
# Find which DOIs are missing
missing_dois = set(doi_txt_list) - set(clean_doi_list)
print(missing_dois)
```
